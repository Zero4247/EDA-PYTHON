```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as mpl
import seaborn as sb
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors
from seaborn import color_palette

Data = pd.read_csv('Superstore.csv', encoding='Latin-1')
```

```python
Data.head(5)
```

```python
Data.info()
```

```python
Data['Order Date'] = pd.to_datetime(Data['Order Date'])
Data['Ship Date'] = pd.to_datetime(Data['Ship Date'])
```

```python
Data.info()
```

```python
Data['Order Month'] = Data['Order Date'].dt.month
Data['Order Year'] = Data['Order Date'].dt.year
Data['Order_Day_Of_Week'] = Data['Order Date'].dt.dayofweek
print(Data)
```

```python
Data.head(5)
```

MONTHLY SALES

```python
sales_by_month= Data.groupby('Order Month')['Sales'].sum().reset_index()
# Customizing labels (maps column names to display names)
labels_dict = {'Order Month': 'Month', 'Sales': 'Revenue ($)'}
fig = px.line(sales_by_month, x='Order Month', y='Sales', title='Sales by Month',
labels=labels_dict)
# Optional: Further layout customization (e.g., center the title)
fig.update_layout(title_x=0.5)
# If you need to show markers on the line as well:
fig.update_traces(mode='lines+markers')
fig.show()
```

```python
# 1. Group the data
sales_by_month = Data.groupby('Order Month')['Sales'].sum().reset_index()
# 2. (Optional) Create a formatted label column if you want currency formatting
# This makes the labels look cleaner (e.g., "$1,200.00")
sales_by_month['Sales_Label'] = sales_by_month['Sales'].apply(lambda x: f"${x:,.2f}")
# 3. Create the plot using the 'text' parameter
fig = px.line(
    sales_by_month,
    x='Order Month',
    y='Sales',
    title='Sales by Month',
    text='Sales_Label'  # This adds the labels
)
# 4. Update the trace to position the labels
fig.update_traces(textposition='top center')
fig.show()
```

SALES BY CATEGORY

```python
sales_by_category= Data.groupby('Category')['Sales'].sum().reset_index()
fig = px.pie(sales_by_category,names='Category',values='Sales',title='Sales by Category')
# 3. Increase height and width
fig.update_layout(
    width=700,   # Set your desired width in pixels
    height=600   # Set your desired height in pixels
)
# 4. Show both Labels and Values together
fig.update_traces(
    textinfo='label+value', # Combines label and value
    textposition='outside'  # Optional: places the text outside the slices for better readability
)
fig.show()
```

SALES BY SUB-CATEGORY

```python
sales_by_subcategory= Data.groupby('Sub-Category')['Sales'].sum().reset_index()
fig = px.pie(sales_by_subcategory,names='Sub-Category',values='Sales',hole=0.4,title='Sales by Sub-Category')
# 3. Increase height and width
fig.update_layout(
    width=700,   # Set your desired width in pixels
    height=600   # Set your desired height in pixels
)
# 4. Show both Labels and Values together
fig.update_traces(
    textinfo='label+value', # Combines label and value
    textposition='outside'  # Optional: places the text outside the slices for better readability
)
fig.show()
```

SALES BY MONTHLY PROFIT

```python
sales_by_monthly_profit= Data.groupby('Order Month')['Profit'].sum().reset_index()
# 2. Create a formatted label column for the values (e.g., "$1,234.56")
sales_by_monthly_profit['Profit_Label'] = sales_by_monthly_profit['Profit'].apply(lambda x: f"${x:,.2f}")
fig = px.bar(sales_by_monthly_profit, x='Order Month', y='Profit', title='Sales by Monthly Profit',text='Profit_Label')
# 4. Increase height & width and update text position
fig.update_layout(
    height=600,  # Adjust height (in pixels)
    width=750   # Adjust width (in pixels)
)
# Position the text on top of the bars
fig.update_traces(textposition='outside')
fig.show()
```

PROFIT BY CATEGORY

```python
sales_by_category_profit= Data.groupby('Category')['Profit'].sum().reset_index()
# 2. Create a formatted label column for the values (e.g., "$1,234.56")
sales_by_category_profit['Profit_Label'] = sales_by_category_profit['Profit'].apply(lambda x: f"${x:,.2f}")
fig = px.bar(sales_by_category_profit, x='Category', y='Profit', title='Sales by Category Profit',text='Profit_Label')
# 4. Increase height & width and update text position
fig.update_layout(
    height=600,  # Adjust height (in pixels)
    width=750   # Adjust width (in pixels)
)
# Position the text on top of the bars
fig.update_traces(textposition='outside')
fig.show()
```

SALES VS PROFIT BY CUSTOMER SEGMENT

```python
# 1. Group the data
sales_vs_profit_by_customer_segment = Data.groupby('Segment').agg({'Sales':'sum', 'Profit':'sum'}).reset_index()
# 2. Define the color palette
color_palette = colors.qualitative.Pastel
# 3. Initialize the figure
fig = go.Figure()
# 4. Add the 'Sales' bar
fig.add_trace(go.Bar(
    x=sales_vs_profit_by_customer_segment['Segment'],
    y=sales_vs_profit_by_customer_segment['Sales'],
    name='Sales',
    marker_color=color_palette[0],
    text=sales_vs_profit_by_customer_segment['Sales'].apply(lambda x: f'${x:,.0f}'), # Add text labels
    textposition='auto' # Automatically place labels
))
# 5. Add the 'Profit' bar
fig.add_trace(go.Bar(
    x=sales_vs_profit_by_customer_segment['Segment'],
    y=sales_vs_profit_by_customer_segment['Profit'],
    name='Profit',
    marker_color=color_palette[1],
    text=sales_vs_profit_by_customer_segment['Profit'].apply(lambda x: f'${x:,.0f}'), # Add text labels
    textposition='auto' # Automatically place labels
))
# 6. Update layout for size, grouping, and labels
fig.update_layout(
    title='Sales vs Profit by Customer Segment',
    xaxis_title='Segment',
    yaxis_title='Amount',
    barmode='group', # Puts bars side-by-side
    height=600,      # Increase height
    width=750       # Increase width
)
fig.show()
```
